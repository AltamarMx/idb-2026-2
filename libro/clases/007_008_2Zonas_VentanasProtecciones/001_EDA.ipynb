```python
import pandas as pd
import matplotlib.pyplot as plt
from iertools.read import read_sql
from dateutil.parser import parse
```

```python
f = "../osm/005_CBs/run/eplusout.sql"
dosz = read_sql(f,alias=True).data
```

```python
f1 = dosz.To.resample("D").mean().idxmax()
```

```python
fig, ax = plt.subplots(figsize=(12,4))
f1 = f1
f2 = f1 + pd.Timedelta("7D")

ax.plot(dosz.Ti_ESTE,label="Ti_este")
ax.plot(dosz.Ti_OESTE,label="Ti_oeste")
ax.plot(dosz.To,label="To")

ax.set_xlim(f1,f2)
ax.legend()
```
